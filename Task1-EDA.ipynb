```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("EDA.ipynb.csv")
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
df.mode().head(1)
```

```python
df.std(numeric_only=True)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)

df["Month"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
monthly = df.groupby("Month")["Sales"].sum()

monthly = monthly.reindex([
"January","February","March","April",
"May","June","July","August",
"September","October","November","December"
])

plt.figure(figsize=(12,5))
monthly.plot(marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
quarter = df.groupby("Quarter")["Sales"].sum()

quarter.plot(kind="bar")

plt.title("Quarterly Sales")

plt.ylabel("Sales")

plt.show()
```

```python
top = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

top.plot(kind="barh")

plt.title("Top 10 Best Selling Products")

plt.xlabel("Sales")

plt.show()
```

```python
category = df.groupby("Category")["Sales"].sum()

category.plot(kind="bar")

plt.title("Revenue by Category")

plt.ylabel("Sales")

plt.show()
```

```python
region = df.groupby("Region")["Sales"].sum()

region.plot(kind="pie",autopct="%1.1f%%")

plt.ylabel("")

plt.title("Sales by Region")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(df.corr(numeric_only=True),
annot=True,
cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Sales"],bins=40)

plt.title("Distribution of Sales")

plt.show()
```
