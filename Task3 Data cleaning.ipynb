```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("train.csv")
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns")
print(df.columns)

print("\nData Types")
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop(columns=["Cabin"])
```

```python
df.isnull().sum()
```

```python
df["Sex"] = df["Sex"].str.capitalize()

df["Embarked"] = df["Embarked"].str.upper()
```

```python
print(df.dtypes)
```

```python
Q1 = df["Fare"].quantile(0.25)

Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[(df["Fare"] < lower) | (df["Fare"] > upper)]

print("Number of Outliers:", len(outliers))
```

```python
before = {
    "Rows": 891,
    "Columns": 12,
    "Missing Values": 866,
    "Duplicate Rows": 0
}

after = {
    "Rows": len(df),
    "Columns": len(df.columns),
    "Missing Values": int(df.isnull().sum().sum()),
    "Duplicate Rows": int(df.duplicated().sum())
}

summary = pd.DataFrame({
    "Before Cleaning": before,
    "After Cleaning": after
})

summary
```

```python
df.to_csv("Titanic_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```

```python
Conclusion

1. Missing values in Age were replaced with the median value.

2. Missing values in Embarked were replaced using the mode.

3. The Cabin column was removed due to excessive missing values.

4. Duplicate rows were checked, and none were found.

5. Text formatting was standardized for consistency.

6. Outliers in Fare were identified using the IQR method and retained because they represent valid passenger fares.

7. The cleaned dataset was saved as Titanic_Cleaned.csv and is ready for further analysis or machine learning.
```
